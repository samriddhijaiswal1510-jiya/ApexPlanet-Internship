# Task 3 — Sales Dashboard
**Objective:** Build an interactive sales dashboard using Power BI, supported by Python-based data preparation and KPI analysis.

**Dataset:** Superstore Sales Dataset

**Tools:** Python (Pandas, Matplotlib, Seaborn), Power BI

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

df = pd.read_csv('train.csv', encoding='latin1')
df['Order Date'] = pd.to_datetime(df['Order Date'], dayfirst=True)
df['Ship Date']  = pd.to_datetime(df['Ship Date'],  dayfirst=True)
df['Year']  = df['Order Date'].dt.year
df['Month'] = df['Order Date'].dt.to_period('M')

print('Shape:', df.shape)
df.head(3)
```

## KPI Summary

```python
total_sales   = df['Sales'].sum()
total_orders  = df['Order ID'].nunique()
avg_order_val = df.groupby('Order ID')['Sales'].sum().mean()
total_customers = df['Customer ID'].nunique()
total_products  = df['Product ID'].nunique()

print(f'Total Sales         : ${total_sales:,.2f}')
print(f'Total Orders        : {total_orders:,}')
print(f'Avg Order Value     : ${avg_order_val:,.2f}')
print(f'Total Customers     : {total_customers:,}')
print(f'Total Products      : {total_products:,}')
```

## Sales by Region

```python
region_sales = df.groupby('Region')['Sales'].sum().sort_values(ascending=False)

plt.figure(figsize=(7, 4))
region_sales.plot(kind='bar', color='steelblue', edgecolor='white')
plt.title('Sales by Region')
plt.xlabel('Region')
plt.ylabel('Total Sales')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
print(region_sales.to_string())
```

## Sales by Segment

```python
seg_sales = df.groupby('Segment')['Sales'].sum().sort_values(ascending=False)

plt.figure(figsize=(6, 4))
seg_sales.plot(kind='bar', color='teal', edgecolor='white')
plt.title('Sales by Segment')
plt.xlabel('Segment')
plt.ylabel('Total Sales')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
print(seg_sales.to_string())
```

## Sales by Category

```python
cat_sales = df.groupby('Category')['Sales'].sum().sort_values(ascending=False)

plt.figure(figsize=(6, 4))
cat_sales.plot(kind='bar', color='coral', edgecolor='white')
plt.title('Sales by Category')
plt.xlabel('Category')
plt.ylabel('Total Sales')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
print(cat_sales.to_string())
```

## Sales Trend Over Time

```python
monthly_sales = df.groupby('Month')['Sales'].sum()
monthly_sales.index = monthly_sales.index.astype(str)

plt.figure(figsize=(14, 4))
plt.plot(monthly_sales.index, monthly_sales.values, marker='o', color='steelblue', linewidth=2)
plt.title('Monthly Sales Trend Over Time')
plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.xticks(rotation=45, ha='right', fontsize=7)
plt.tight_layout()
plt.show()
```

## Top 10 Products by Sales

```python
top_products = df.groupby('Product Name')['Sales'].sum().sort_values(ascending=False).head(10)

plt.figure(figsize=(10, 5))
top_products.sort_values().plot(kind='barh', color='steelblue', edgecolor='white')
plt.title('Top 10 Products by Sales')
plt.xlabel('Total Sales')
plt.tight_layout()
plt.show()
print(top_products.to_string())
```
